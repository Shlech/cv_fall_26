# ЛР3 (усложнённая): SR-inpainting на основе SUPIR

Сэмплинг SUPIR изменён так, что восстанавливается только фрагмент изображения, заданный бинарной маской
(маска задаётся в разрешении **входного** LR-изображения). Вне маски известно изображение в высоком качестве (HR):
на каждом шаге диффузии латент вне маски заменяется зашумлённым латентом HR, а в конце HR вклеивается по маске.

Сравниваются два режима восстановления одной и той же области:
* `M ≡ 1` — у модели есть только LR (обычный SUPIR);
* `combined` — внутри маски модель восстанавливает изображение по LR, но может опираться на HR-фрагменты вокруг.

Разница должна быть особенно заметна на сильно испорченном входе, поэтому к деградациям DIV2K добавлена `heavy`.

**Как запустить на Kaggle**
1. *Settings → Accelerator*: **GPU T4 x2**. *Settings → Internet*: **On**.
2. *Run All*. Скачивание весов (~10 ГБ) и DIV2K (~0.5 ГБ) — около 10–15 минут, эксперимент с абляцией — около полутора–двух часов при параметрах по умолчанию.
3. Таблицы, графики и изображения складываются в `/kaggle/working/results` (раскладка — как в папке `results/`
   репозитория: `hr/`, `lq/`, `outputs/`, `grids/`, `zoom/`, `ablation/`, `chart.png`, `metrics.csv`),
   а также в архив `/kaggle/working/results.zip` (вкладка *Output*). Отчёт — `README.md` рядом с ноутбуком.

SUPIR рассчитан на работу при короткой стороне ≥ 1024 px (параметр `min_size` в его `test.py`): на меньшем
разрешении SDXL рисует повторяющиеся текстуры («сетку», муар). Поэтому LQ увеличивается до 1024, обрабатывается,
а результат уменьшается до выходного разрешения x4. Модель разнесена по двум GPU: UNet + адаптер на `cuda:0`,
VAE, текстовые энкодеры и метрики — на `cuda:1` (на одной T4 при 1024 не хватает памяти).

Если упадёт по памяти GPU — поставьте `USE_TILE_VAE = True` или уменьшите `PROC_MIN_SIZE` до 768.
Если упадёт по ОЗУ при загрузке модели — перезапустите сессию (*Restart*) и запустите заново.

## Данные: DIV2K valid, четыре вида деградаций

* `bicubic` — официальные DIV2K LR x4 (bicubic);
* `unknown` — официальные DIV2K LR x4 «unknown» (неизвестное ядро);
* `realesrgan` — синтетическая деградация второго порядка по схеме Real-ESRGAN (размытие → ресайз → шум → JPEG, дважды;
  sinc-фильтр опущен), генерируется из HR с фиксированным seed;
* `heavy` — сильная деградация: гауссово размытие σ=2.5 → уменьшение x4 → гауссов шум σ=20/255 → JPEG q=20.

```python
DIV2K_URL = 'http://data.vision.ee.ethz.ch/cvl/DIV2K/'
if DIV2K_ROOT is None:
    DIV2K_ROOT = DATA_DIR
    for name in ['DIV2K_valid_HR', 'DIV2K_valid_LR_bicubic_X4', 'DIV2K_valid_LR_unknown_X4']:
        zp, done = f'{DATA_DIR}/{name}.zip', f'{DATA_DIR}/.{name}.done'
        if os.path.exists(done):
            continue
        print('downloading', name)
        subprocess.run(['wget', '-q', '-c', '-O', zp, DIV2K_URL + name + '.zip'], check=True)
        with zipfile.ZipFile(zp) as z:
            z.extractall(DATA_DIR)
        os.remove(zp)
        open(done, 'w').close()

hr_files = sorted(glob.glob(os.path.join(DIV2K_ROOT, '**', 'DIV2K_valid_HR', '*.png'), recursive=True))
assert hr_files, f'Не найдены HR-изображения DIV2K в {DIV2K_ROOT}'

def lr_path(kind, img_id):
    hits = glob.glob(os.path.join(DIV2K_ROOT, '**', f'DIV2K_valid_LR_{kind}', '**', f'{img_id}x4.png'), recursive=True)
    assert hits, f'Не найден LR ({kind}) для {img_id}'
    return hits[0]

print(len(hr_files), 'HR images')
```

```python
import cv2
from PIL import Image

def _jpeg(img, q):
    u8 = (np.clip(img, 0, 1) * 255).round().astype(np.uint8)
    ok, buf = cv2.imencode('.jpg', cv2.cvtColor(u8, cv2.COLOR_RGB2BGR), [cv2.IMWRITE_JPEG_QUALITY, int(q)])
    return cv2.cvtColor(cv2.imdecode(buf, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB).astype(np.float32) / 255

def _blur(img, rng, sig_range, ksize=21):
    sx, sy = rng.uniform(*sig_range), rng.uniform(*sig_range)
    if rng.random() < 0.5:                       # изотропное ядро
        sy = sx
    theta = rng.uniform(0, np.pi)
    ax = np.arange(ksize) - ksize // 2
    xx, yy = np.meshgrid(ax, ax)
    xr = np.cos(theta) * xx + np.sin(theta) * yy
    yr = -np.sin(theta) * xx + np.cos(theta) * yy
    k = np.exp(-0.5 * (xr ** 2 / sx ** 2 + yr ** 2 / sy ** 2))
    return cv2.filter2D(img, -1, (k / k.sum()).astype(np.float32), borderType=cv2.BORDER_REFLECT)

def _resize(img, rng, scale_range):
    mode = rng.choice(['up', 'down', 'keep'], p=[0.2, 0.7, 0.1])
    s = rng.uniform(1, scale_range[1]) if mode == 'up' else rng.uniform(scale_range[0], 1) if mode == 'down' else 1.0
    interp = int(rng.choice([cv2.INTER_AREA, cv2.INTER_LINEAR, cv2.INTER_CUBIC]))
    h, w = img.shape[:2]
    return cv2.resize(img, (max(8, int(w * s)), max(8, int(h * s))), interpolation=interp)

def _noise(img, rng, gauss_range, poisson_range):
    if rng.random() < 0.5:                       # гауссов шум (цветной или серый)
        sigma = rng.uniform(*gauss_range) / 255
        shape = img.shape[:2] + (1,) if rng.random() < 0.4 else img.shape
        return img + rng.normal(0, sigma, shape).astype(np.float32)
    clean = np.clip(img, 0, 1)                   # пуассоновский шум
    noisy = rng.poisson(clean * 255).astype(np.float32) / 255
    return img + (noisy - clean) * rng.uniform(*poisson_range)

def realesrgan_degrade(hr_u8, scale=4, seed=0):
    """Упрощённая деградация второго порядка Real-ESRGAN (Wang et al., 2021)."""
    rng = np.random.default_rng(seed)
    h, w = hr_u8.shape[:2]
    img = hr_u8.astype(np.float32) / 255
    # порядок 1
    img = _blur(img, rng, (0.2, 3.0))
    img = _resize(img, rng, (0.15, 1.5))
    img = _noise(img, rng, (1, 30), (0.05, 3.0))
    img = _jpeg(img, rng.uniform(30, 95))
    # порядок 2
    if rng.random() < 0.8:
        img = _blur(img, rng, (0.2, 1.5))
    img = _resize(img, rng, (0.3, 1.2))
    img = _noise(img, rng, (1, 25), (0.05, 2.5))
    # финал: приведение к целевому размеру + JPEG
    img = cv2.resize(np.clip(img, 0, 1), (w // scale, h // scale),
                     interpolation=int(rng.choice([cv2.INTER_AREA, cv2.INTER_LINEAR, cv2.INTER_CUBIC])))
    img = _jpeg(img, rng.uniform(30, 95))
    return (img * 255).round().clip(0, 255).astype(np.uint8)

def heavy_degrade(hr_u8, scale=4, seed=0):
    """Сильная фиксированная деградация: размытие σ=2.5 → x1/scale → шум σ=20/255 → JPEG q=20."""
    rng = np.random.default_rng(seed)
    h, w = hr_u8.shape[:2]
    img = cv2.GaussianBlur(hr_u8.astype(np.float32) / 255, (0, 0), 2.5, borderType=cv2.BORDER_REFLECT)
    img = cv2.resize(img, (w // scale, h // scale), interpolation=cv2.INTER_AREA)
    img = img + rng.normal(0, 20 / 255, img.shape).astype(np.float32)
    img = _jpeg(img, 20)
    return (img * 255).round().clip(0, 255).astype(np.uint8)

def load_rgb(path):
    return np.array(Image.open(path).convert('RGB'))
```

```python
import matplotlib.pyplot as plt

ids = IMAGE_IDS or [os.path.splitext(os.path.basename(p))[0] for p in hr_files[:N_IMAGES]]
hr_by_id = {os.path.splitext(os.path.basename(p))[0]: p for p in hr_files}

samples = []
for img_id in ids:
    hr = load_rgb(hr_by_id[img_id])
    H, W = hr.shape[:2]
    hc = LR_CROP * SCALE
    # центральный фрагмент, выровненный по сетке x4, чтобы LR и HR совпадали
    y0 = (H - hc) // 2 // SCALE * SCALE
    x0 = (W - hc) // 2 // SCALE * SCALE
    hr_crop = hr[y0:y0 + hc, x0:x0 + hc]
    ly, lx = y0 // SCALE, x0 // SCALE
    for deg in DEGRADATIONS:
        if deg == 'realesrgan':
            lr = realesrgan_degrade(hr_crop, SCALE, seed=int(img_id))
        elif deg == 'heavy':
            lr = heavy_degrade(hr_crop, SCALE, seed=int(img_id))
        else:
            lr = load_rgb(lr_path(deg, img_id))[ly:ly + LR_CROP, lx:lx + LR_CROP]
        assert lr.shape[:2] == (LR_CROP, LR_CROP), (img_id, deg, lr.shape)
        samples.append(dict(img_id=img_id, deg=deg, hr=hr_crop, lr=lr))

fig, axes = plt.subplots(len(ids), len(DEGRADATIONS) + 1, figsize=(3 * (len(DEGRADATIONS) + 1), 3 * len(ids)), squeeze=False)
for r, img_id in enumerate(ids):
    row = [s for s in samples if s['img_id'] == img_id]
    axes[r, 0].imshow(row[0]['hr']); axes[r, 0].set_title(f'{img_id} HR')
    for c, s in enumerate(row, 1):
        axes[r, c].imshow(s['lr'], interpolation='nearest'); axes[r, c].set_title(f'LR {s["deg"]}')
for a in axes.flat:
    a.axis('off')
plt.tight_layout(); plt.savefig(f'{OUT_DIR}/inputs.png', dpi=100); plt.show()
```

## Маски
Маска — бинарный массив `uint8` размера входного LR-изображения (`LR_CROP x LR_CROP`).
`M = 1` — область, которую модель восстанавливает по LR; `M = 0` — область, где известен HR.
`combined` — эллипс в центре (≈ треть площади): внутри — восстановление по LR, снаружи — HR.

```python
def make_mask(kind, h, w):
    if kind == 'ones':
        return np.ones((h, w), np.uint8)
    if kind == 'zeros':
        return np.zeros((h, w), np.uint8)
    if kind == 'combined':
        yy, xx = np.mgrid[:h, :w]
        return ((((yy - h / 2) / (0.32 * h)) ** 2 + ((xx - w / 2) / (0.32 * w)) ** 2) <= 1).astype(np.uint8)
    raise ValueError(kind)

MASKS = {k: make_mask(k, LR_CROP, LR_CROP) for k in MASK_KINDS}
ROI = make_mask('combined', LR_CROP, LR_CROP)   # по этой области считаем метрики «внутри/снаружи» для всех масок

fig, axes = plt.subplots(1, len(MASKS), figsize=(3 * len(MASKS), 3))
for a, (k, m) in zip(axes, MASKS.items()):
    a.imshow(m, cmap='gray', vmin=0, vmax=1); a.set_title(f'{k}: {m.mean():.0%}'); a.axis('off')
plt.tight_layout(); plt.savefig(f'{OUT_DIR}/masks.png', dpi=100); plt.show()
```

## Загрузка SUPIR

* xformers не ставим (он потянул бы другую версию torch) — в конфиге переключаем внимание на стандартное (`torch` SDPA).
* Модель создаётся в fp32 на CPU (~19 ГБ ОЗУ), сразу переводится в fp16, затем загружаются веса SDXL и SUPIR.
* VAE оставляем в fp32: в fp16 VAE SDXL даёт NaN, а bf16 у T4 нет.
* UNet + адаптер — на `DEVICE`, VAE (с деградационно-устойчивым энкодером) и текстовые энкодеры — на `AE_DEVICE`.

```python
os.chdir(SUPIR_DIR)

# Текстовый энкодер OpenCLIP ViT-bigG: его веса лежат в чекпоинте SDXL, поэтому
#  1) не скачиваем веса LAION отдельно (~10 ГБ);
#  2) создаём модель с крошечной визуальной частью — SGM всё равно сразу её удаляет, а полноразмерная
#     визуальная башня bigG в fp32 заняла бы ещё ~7 ГБ ОЗУ в пике и не влезла бы в лимит Kaggle.
from open_clip.factory import _MODEL_CONFIGS
_text_only_cfg = copy.deepcopy(_MODEL_CONFIGS['ViT-bigG-14'])
_text_only_cfg['vision_cfg'].update(layers=1, width=_text_only_cfg['vision_cfg'].get('head_width', 64))
_MODEL_CONFIGS['ViT-bigG-14-textonly'] = _text_only_cfg

# Оригинал берём из open_clip.factory: подменяем только атрибут пакета, так что там он всегда настоящий
# (и ячейку можно безопасно перезапускать).
import open_clip.factory
def _create_text_only(model_name, *args, **kwargs):
    kwargs['pretrained'] = None
    if model_name == 'ViT-bigG-14':
        model_name = 'ViT-bigG-14-textonly'
    return open_clip.factory.create_model_and_transforms(model_name, *args, **kwargs)
open_clip.create_model_and_transforms = _create_text_only

# sgm/modules/diffusionmodules/sampling.py импортирует k_diffusion только ради DPM++-сэмплеров, которые
# SUPIR здесь не использует. Пакет k-diffusion тянет wandb, clean-fid и т.п., поэтому вместо него — заглушка.
try:
    import k_diffusion.sampling
except ImportError:
    import types
    def _k_diffusion_unused(*args, **kwargs):
        raise NotImplementedError('k_diffusion не установлен: этот сэмплер в ноутбуке не используется')
    _kd, _kds = types.ModuleType('k_diffusion'), types.ModuleType('k_diffusion.sampling')
    _kds.get_sigmas_karras = _kds.BrownianTreeNoiseSampler = _k_diffusion_unused
    _kd.sampling = _kds
    sys.modules['k_diffusion'], sys.modules['k_diffusion.sampling'] = _kd, _kds

from omegaconf import OmegaConf
from sgm.util import instantiate_from_config
from SUPIR.util import load_state_dict
from SUPIR.utils.colorfix import wavelet_reconstruction
from sgm.modules.diffusionmodules.sampling import RestoreEDMSampler
from sgm.modules.distributions.distributions import DiagonalGaussianDistribution

def _no_xformers(o):
    if isinstance(o, dict):
        return {k: _no_xformers(v) for k, v in o.items()}
    if isinstance(o, list):
        return [_no_xformers(v) for v in o]
    return {'vanilla-xformers': 'vanilla', 'softmax-xformers': 'softmax'}.get(o, o) if isinstance(o, str) else o

cfg = OmegaConf.load(f'{SUPIR_DIR}/options/SUPIR_v0.yaml')
model_cfg = OmegaConf.create(_no_xformers(OmegaConf.to_container(cfg.model)))
model_cfg.params.ae_dtype = 'fp32'

t0 = time.time()
if 'model' in globals():                 # при перезапуске ячейки освобождаем старую модель
    del model
    gc.collect(); torch.cuda.empty_cache()
model = instantiate_from_config(model_cfg).cpu()
model = model.half()
gc.collect()

model_keys = set(model.state_dict().keys())
loaded_keys = set()
for ckpt in (SDXL_CKPT, SUPIR_CKPT):
    sd = load_state_dict(ckpt)
    # в transformers 5.x у CLIPTextModel нет вложенного text_model — переименовываем ключи CLIP ViT-L из SDXL
    for k in [k for k in sd if k not in model_keys and '.text_model.' in k]:
        k2 = k.replace('.text_model.', '.', 1)
        if k2 in model_keys:
            sd[k2] = sd.pop(k)
    matched = [k for k in sd if k in model_keys]
    print(f'{os.path.basename(ckpt)}: совпало {len(matched)} / {len(sd)} ключей')
    unexpected = [k for k in sd if k not in model_keys]
    if unexpected:
        print('   лишние ключи (пример):', unexpected[:3])
    assert len(matched) > 0.9 * len(sd), 'Чекпоинт не подходит к модели'
    model.load_state_dict(sd, strict=False)
    loaded_keys |= set(matched)
    del sd; gc.collect()
missing = sorted(model_keys - loaded_keys)
print(f'Параметров модели без весов из чекпоинтов: {len(missing)}', missing[:5])

model.first_stage_model.float()          # VAE (+ деградационно-устойчивый энкодер) в fp32
model.ae_dtype = torch.float32
model.model.dtype = torch.float16        # UNet + адаптер в fp16 через autocast
model.eval()
for name, child in model.named_children():
    child.to(AE_DEVICE if name in ('first_stage_model', 'conditioner') else DEVICE)
# Текстовые энкодеры SGM хранят строку self.device='cuda' (= cuda:0) и переносят на неё токены — исправляем
for m in model.conditioner.modules():
    if isinstance(vars(m).get('device'), str):
        m.device = AE_DEVICE
if USE_TILE_VAE is None:
    USE_TILE_VAE = AE_DEVICE == DEVICE
if USE_TILE_VAE:
    model.init_tile_vae(encoder_tile_size=512, decoder_tile_size=64)
gc.collect(); torch.cuda.empty_cache()

def gpu_mem(fn=torch.cuda.memory_allocated):
    return ', '.join(f'{d}: {fn(d) / 2**30:.1f} GB' for d in dict.fromkeys([DEVICE, AE_DEVICE]))

print(f'Модель загружена за {time.time() - t0:.0f} с; тайловый VAE: {USE_TILE_VAE}; GPU — {gpu_mem()}')
```

## Масочный сэмплер (SR-inpainting)

Исходный `RestoreEDMSampler` SUPIR — это EDM/Euler-сэмплер (Karras et al.) в латенте SDXL с
restoration-guided поправкой. Изменения:

1. Маска `M` (LR-разрешение) → `nearest` до выходного разрешения `M_pix` → `max_pool 8x8` до латента `m`
   (+ дилатация на `LATENT_DILATE` латентных пикселей, чтобы модель видела контекст на границе).
2. Известный латент `z_known = E(x_known)` — латент известного изображения высокого качества (HR,
   увеличенного до рабочего разрешения SUPIR). LR по-прежнему подаётся в SUPIR как условие (адаптер, `x_center`).
3. После **каждого** шага Эйлера (и перед первым) латент смешивается:
   `x_{σ_{i+1}} ← m ⊙ x_{σ_{i+1}} + (1 − m) ⊙ (z_known + σ_{i+1}·ε)`, где `ε` — фиксированный шум.
   Вне маски траектория всё время совпадает с «зашумлённым HR», поэтому модель достраивает
   область маски согласованно с настоящим окружением (идея Blended Latent Diffusion / RePaint).
   На последнем шаге `σ = 0` ⇒ вне маски латент ровно `z_known`.
   Параметр `blend_frac` ограничивает смешивание первыми `blend_frac·N` шагами: дальше латент вне маски
   развивается свободно (вне маски выход всё равно восстанавливается шагом 4). Исследуется в абляции ниже.
4. После декодирования — финальное смешивание в пикселях с размытой маской:
   `y = M̃ ⊙ D(x_0) + (1 − M̃) ⊙ x_known`, чтобы вне маски не было даже потерь VAE.

При `M ≡ 1` алгоритм в точности совпадает с исходным SUPIR (HR не используется), при `M ≡ 0` выход равен `x_known`.

```python
import torch.nn.functional as F
import torchvision.transforms.functional as TF

class MaskedRestoreEDMSampler(RestoreEDMSampler):
    """RestoreEDMSampler из SUPIR + латентное смешивание по маске."""

    def __call__(self, denoiser, x, cond, uc=None, num_steps=None, x_center=None, control_scale=1.0,
                 use_linear_control_scale=False, control_scale_start=0.0,
                 mask=None, z_known=None, known_noise=None, blend_frac=1.0):
        x, s_in, sigmas, num_sigmas, cond, uc = self.prepare_sampling_loop(x, cond, uc, num_steps)
        n_blend = round(blend_frac * (num_sigmas - 1))   # смешивание только на первых n_blend шагах

        def blend(x, sigma):
            if mask is None:
                return x
            return mask * x + (1 - mask) * (z_known + sigma * known_noise)

        if n_blend > 0:
            x = blend(x, sigmas[0])
        for i in self.get_sigma_gen(num_sigmas):
            gamma = (min(self.s_churn / (num_sigmas - 1), 2 ** 0.5 - 1)
                     if self.s_tmin <= sigmas[i] <= self.s_tmax else 0.0)
            x = self.sampler_step(s_in * sigmas[i], s_in * sigmas[i + 1], denoiser, x, cond, uc, gamma, x_center,
                                  control_scale=control_scale,
                                  use_linear_control_scale=use_linear_control_scale,
                                  control_scale_start=control_scale_start)
            if i < n_blend:
                x = blend(x, sigmas[i + 1])
        return x


def mask_to_pixel(mask_lr, H, W, device=AE_DEVICE):
    m = torch.from_numpy(np.asarray(mask_lr, dtype=np.float32))[None, None].to(device)
    return F.interpolate(m, size=(H, W), mode='nearest')

def mask_to_latent(m_pix, dilate=LATENT_DILATE):
    m = F.max_pool2d(m_pix, 8)
    if dilate > 0:
        m = F.max_pool2d(m, 2 * dilate + 1, stride=1, padding=dilate)
    return m

def feather(m_pix, sigma=FEATHER_SIGMA):
    if sigma <= 0:
        return m_pix
    k = int(4 * sigma) * 2 + 1
    return TF.gaussian_blur(m_pix, kernel_size=k, sigma=sigma)

@torch.no_grad()
def encode_mode(x):
    """Латент VAE без случайности (мода апостериорного распределения)."""
    fsm = model.first_stage_model
    return model.scale_factor * DiagonalGaussianDistribution(fsm.quant_conv(fsm.encoder(x))).mode()


def _to_dev(d, device):
    return {k: (v.to(device) if torch.is_tensor(v) else v) for k, v in d.items()}

@torch.no_grad()
def supir_sr_inpaint(x_up, mask_lr=None, x_known=None, num_steps=STEPS, seed=SEED, p='', p_p=A_PROMPT, n_p=N_PROMPT,
                     cfg_scale=7.5, cfg_scale_start=4.0, use_linear_CFG=True, restoration_scale=-1,
                     s_churn=5, s_noise=1.003, control_scale=1.0, color_fix_type='Wavelet',
                     feather_sigma=FEATHER_SIGMA, blend_frac=1.0):
    """
    x_up    : [1,3,H,W] в [-1,1] на AE_DEVICE — LQ, уже увеличенный до рабочего разрешения SUPIR (условие модели).
    mask_lr : np.ndarray (h,w) из {0,1} в разрешении входного LR; 1 — восстанавливать, 0 — известная область;
              None — исходный SUPIR.
    x_known : [1,3,H,W] — известное изображение для области M=0 (HR); None — сам x_up.
    Повторяет SUPIRModel.batchify_sample, но с MaskedRestoreEDMSampler.
    VAE и текстовые энкодеры работают на AE_DEVICE, диффузия — на DEVICE.
    """
    H, W = x_up.shape[-2:]
    if x_known is None:
        x_known = x_up

    # --- маска и известная область (до установки seed, чтобы не сдвигать RNG исходного алгоритма)
    if mask_lr is not None:
        m_pix = mask_to_pixel(mask_lr, H, W, device=x_up.device)
        m_lat = mask_to_latent(m_pix).to(DEVICE)
        z_known = encode_mode(x_known).to(DEVICE)
        g = torch.Generator(device=DEVICE).manual_seed(seed + 1)
        known_noise = torch.randn(z_known.shape, generator=g, device=DEVICE, dtype=z_known.dtype)
    else:
        m_pix = m_lat = z_known = known_noise = None

    # --- сэмплер с теми же настройками, что batchify_sample
    sc = copy.deepcopy(model.sampler_config)
    sc.params.num_steps = num_steps
    sc.params.guider_config.params.scale_min = cfg_scale
    sc.params.guider_config.params.scale = cfg_scale_start if use_linear_CFG else cfg_scale
    sc.params.restore_cfg = restoration_scale
    sc.params.s_churn = s_churn
    sc.params.s_noise = s_noise
    sampler = MaskedRestoreEDMSampler(**OmegaConf.to_container(sc.params))

    torch.manual_seed(seed)
    _z = model.encode_first_stage_with_denoise(x_up, use_sample=False)
    x_stage1 = model.decode_first_stage(_z)
    z_stage1 = model.encode_first_stage(x_stage1)

    ae_dtype, model.ae_dtype = model.ae_dtype, torch.float16   # текстовые энкодеры — под autocast fp16
    try:
        c, uc = model.prepare_condition(_z, [p], p_p, n_p, 1)
    finally:
        model.ae_dtype = ae_dtype
    c, uc = _to_dev(c, DEVICE), _to_dev(uc, DEVICE)
    _z, z_stage1 = _z.to(DEVICE), z_stage1.to(DEVICE)

    denoiser = lambda inp, sigma, c, control_scale: model.denoiser(model.model, inp, sigma, c, control_scale)
    noised_z = torch.randn_like(_z)
    z = sampler(denoiser, noised_z, cond=c, uc=uc, x_center=z_stage1, control_scale=control_scale,
                mask=m_lat, z_known=z_known, known_noise=known_noise, blend_frac=blend_frac)
    del c, uc, _z, z_stage1, noised_z

    out = model.decode_first_stage(z.float().to(x_up.device))
    if color_fix_type == 'Wavelet':
        out = wavelet_reconstruction(out, x_stage1)
    if m_pix is not None:
        m_soft = feather(m_pix, feather_sigma)
        out = m_soft * out + (1 - m_soft) * x_known
    return out.clamp(-1, 1)
```

### Полный пайплайн: LR + маска + HR → выход x4
1. `x_out` — известное изображение в выходном разрешении: HR (если HR не передан — бикубический апскейл LR).
2. Если маска пустая (`M ≡ 0`), ответ — `x_out`, сэмплирование не запускается.
3. Иначе, как в `SUPIR/test.py` (`PIL2Tensor` с `min_size=1024`), LR бикубически увеличивается до рабочего
   разрешения SUPIR (короткая сторона ≥ `PROC_MIN_SIZE`, кратно 64) — это вход модели; HR увеличивается до того же
   размера — это известная область. Обработка — `supir_sr_inpaint`.
4. Результат уменьшается до выходного разрешения (бикубика с антиалиасингом) и ещё раз смешивается с `x_out`
   по размытой маске — так вне маски выход в точности совпадает с HR.

```python
def to_tensor(img_u8, device=AE_DEVICE):
    return torch.from_numpy(img_u8).permute(2, 0, 1)[None].float().div(127.5).sub(1).to(device)

def to_uint8(t):
    return ((t[0].clamp(-1, 1) + 1) * 127.5).round().byte().permute(1, 2, 0).cpu().numpy()

def resize_bicubic(img_u8, h, w):
    return np.array(Image.fromarray(img_u8).resize((w, h), Image.BICUBIC))

def upscale_bicubic(lr_u8, scale=SCALE):
    h, w = lr_u8.shape[:2]
    return resize_bicubic(lr_u8, h * scale, w * scale)

def proc_size(h, w, min_size=PROC_MIN_SIZE):
    k = max(1.0, min_size / min(h, w))
    return int(round(h * k / 64)) * 64, int(round(w * k / 64)) * 64

def prepare_lq(lr_u8):
    """LQ для SUPIR: бикубика до рабочего разрешения (как PIL2Tensor в SUPIR/test.py)."""
    h, w = lr_u8.shape[:2]
    return to_tensor(resize_bicubic(lr_u8, *proc_size(h * SCALE, w * SCALE)))

@torch.no_grad()
def sr_inpaint(lr_u8, mask_lr=None, hr_u8=None, **kw):
    """LR (uint8, h×w) + маска (h×w) + HR (uint8, h*SCALE×w*SCALE) -> выход [1,3,h*SCALE,w*SCALE] в [-1,1]
    на AE_DEVICE. Внутри маски — восстановление по LR, вне маски — HR."""
    known_u8 = hr_u8 if hr_u8 is not None else upscale_bicubic(lr_u8)
    x_out = to_tensor(known_u8)
    if mask_lr is not None and not np.any(mask_lr):
        return x_out                                       # M ≡ 0: восстанавливать нечего
    H, W = x_out.shape[-2:]
    x_proc = prepare_lq(lr_u8)
    x_known = to_tensor(resize_bicubic(known_u8, *x_proc.shape[-2:]))
    k = x_proc.shape[-2] / H
    y = supir_sr_inpaint(x_proc, mask_lr, x_known=x_known, feather_sigma=FEATHER_SIGMA * k, **kw)
    if y.shape[-2:] != (H, W):
        y = F.interpolate(y, size=(H, W), mode='bicubic', align_corners=False, antialias=True)
    if mask_lr is not None and not np.all(mask_lr):
        m_soft = feather(mask_to_pixel(mask_lr, H, W, device=y.device))
        y = m_soft * y + (1 - m_soft) * x_out
    return y.clamp(-1, 1)
```

## Проверка корректности
1. `M ≡ 1` должен совпасть с исходным SUPIR (`mask=None`) при одинаковом seed — HR при этом не используется.
2. `M ≡ 0`, пропущенная через масочный сэмплер (без короткого пути в `sr_inpaint`), должна в точности дать
   известное изображение (HR в рабочем разрешении).

```python
s0 = samples[0]
for d in dict.fromkeys([DEVICE, AE_DEVICE]):
    torch.cuda.reset_peak_memory_stats(d)
t0 = time.time(); y_ref = sr_inpaint(s0['lr'], None, s0['hr'], **SAMPLER_KW); t_ref = time.time() - t0
y_one = sr_inpaint(s0['lr'], MASKS['ones'], s0['hr'], **SAMPLER_KW)
x_proc0 = prepare_lq(s0['lr'])
x_known0 = to_tensor(resize_bicubic(s0['hr'], *x_proc0.shape[-2:]))
y_zero = supir_sr_inpaint(x_proc0, MASKS['zeros'], x_known=x_known0, **SAMPLER_KW)
d1 = (y_ref - y_one).abs().max().item()
d0 = (y_zero - x_known0).abs().max().item()
print(f'Рабочее разрешение SUPIR: {tuple(x_proc0.shape[-2:])}, выход: {tuple(y_ref.shape[-2:])}')
print(f'Один прогон: {t_ref:.1f} с; пик памяти — {gpu_mem(torch.cuda.max_memory_allocated)}')
print(f'max|SUPIR - mask≡1| = {d1:.2e}   max|HR - mask≡0| = {d0:.2e}')
assert d0 == 0.0, 'mask≡0 должен давать ровно известное изображение'
if d1 >= 2 / 255:
    print('ВНИМАНИЕ: mask≡1 заметно отличается от исходного SUPIR — проверьте детерминизм GPU / код сэмплера')
del y_zero, x_proc0, x_known0

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for a, (t, name) in zip(axes, [(to_tensor(upscale_bicubic(s0['lr'])), 'bicubic'), (y_ref, 'SUPIR'), (y_one, 'mask≡1')]):
    a.imshow(to_uint8(t)); a.set_title(name); a.axis('off')
plt.tight_layout(); plt.show()
```

## Метрики
Главная область сравнения — внутри эллипса `ROI` (он совпадает с маской `combined`): там восстанавливают все методы,
а вне него у `combined` просто HR.
* **PSNR** — по Y-каналу (YCbCr, BT.601), как принято в SR; внутри ROI (`in`), в полосе ±`EDGE_PX` px вокруг
  границы (`edge`, качество шва), снаружи (`out`) и по всему изображению (`full`).
* **LPIPS** (AlexNet) — пространственная карта, усреднённая по тем же областям.
* **CLIP-IQA** (pyiqa, без эталона) — по всему изображению и по описанному вокруг ROI квадрату (`clipiqa_in`);
  для справки считается и для HR.

```python
import lpips, pyiqa

lpips_fn = lpips.LPIPS(net='alex', spatial=True, verbose=False).to(AE_DEVICE).eval()
clipiqa_fn = pyiqa.create_metric('clipiqa', device=AE_DEVICE)

_ry, _rx = np.nonzero(ROI)
ROI_BOX = (_ry.min() * SCALE, (_ry.max() + 1) * SCALE, _rx.min() * SCALE, (_rx.max() + 1) * SCALE)   # y0, y1, x0, x1

def rgb_to_y(x01):
    r, g, b = x01[:, 0:1], x01[:, 1:2], x01[:, 2:3]
    return (16 + 65.481 * r + 128.553 * g + 24.966 * b) / 255

def masked_psnr(a01, b01, w):
    mse = (((rgb_to_y(a01) - rgb_to_y(b01)) ** 2) * w).sum() / w.sum()
    return float(10 * torch.log10(1 / mse.clamp_min(1e-12)))

def edge_band(roi_pix, r=EDGE_PX):
    """Полоса шириной 2r вокруг границы области (для оценки шва)."""
    k = 2 * r + 1
    dil = F.max_pool2d(roi_pix, k, stride=1, padding=r)
    ero = 1 - F.max_pool2d(1 - roi_pix, k, stride=1, padding=r)
    return dil - ero

def clipiqa_scores(x01):
    y0, y1, x0, x1 = ROI_BOX
    x01 = x01.clamp(0, 1)
    return float(clipiqa_fn(x01)), float(clipiqa_fn(x01[..., y0:y1, x0:x1].contiguous()))

@torch.no_grad()
def compute_metrics(out, hr, roi_pix):
    o01, h01 = (out + 1) / 2, (hr + 1) / 2
    lp = lpips_fn(out, hr)                                 # [1,1,H,W]
    res = {}
    regions = {'full': torch.ones_like(roi_pix), 'in': roi_pix, 'out': 1 - roi_pix, 'edge': edge_band(roi_pix)}
    for name, w in regions.items():
        res[f'psnr_{name}'] = masked_psnr(o01, h01, w)
        res[f'lpips_{name}'] = float((lp * w).sum() / w.sum())
    res['clipiqa'], res['clipiqa_in'] = clipiqa_scores(o01)
    return res
```

## Эксперимент: изображения × деградации × маски
Для каждого LR: бикубика (опорный метод), `M ≡ 1` (только LR), `M ≡ 0` (выход = HR, проверяется),
`combined` (восстановление внутри маски по LR с опорой на HR снаружи).

```python
import pandas as pd

rows, hr_iqa, hr_iqa_in, main_out, zero_err = [], {}, {}, {}, []
for si, s in enumerate(samples):
    key = f"{s['img_id']}_{s['deg']}"
    hr_t = to_tensor(s['hr'])
    roi_pix = mask_to_pixel(ROI, *hr_t.shape[-2:])
    if s['img_id'] not in hr_iqa:
        hr_iqa[s['img_id']], hr_iqa_in[s['img_id']] = clipiqa_scores((hr_t + 1) / 2)
        Image.fromarray(s['hr']).save(f'{OUT_DIR}/hr/{s["img_id"]}.png')
    Image.fromarray(s['lr']).save(f'{OUT_DIR}/lq/{key}.png')
    outs = {'bicubic': (to_tensor(upscale_bicubic(s['lr'])), 0.0)}
    for mk in MASK_KINDS:
        t0 = time.time()
        y = sr_inpaint(s['lr'], MASKS[mk], s['hr'], **SAMPLER_KW)
        outs[mk] = (y, time.time() - t0)
    zero_err.append((outs.pop('zeros')[0] - hr_t).abs().max().item())   # M≡0: выход должен быть ровно HR
    for kind, (y, dt) in outs.items():
        Image.fromarray(to_uint8(y)).save(f'{OUT_DIR}/outputs/{key}_{kind}.png')
        main_out[(key, kind)] = y.cpu()
        rows.append(dict(img_id=s['img_id'], degradation=s['deg'], mask=kind, time_s=dt,
                         **compute_metrics(y, hr_t, roi_pix)))
        r = rows[-1]
        print(f'[{si + 1}/{len(samples)}] {key:<16} {kind:<9} {dt:5.1f}s  PSNR in {r["psnr_in"]:.2f}  '
              f'LPIPS in {r["lpips_in"]:.3f}  CLIP-IQA in {r["clipiqa_in"]:.3f}')
    del outs
    torch.cuda.empty_cache()

d0_hr = max(zero_err)
print(f'max|выход(M≡0) - HR| по всем изображениям: {d0_hr:.2e}')
assert d0_hr == 0.0
df = pd.DataFrame(rows)
df.to_csv(f'{OUT_DIR}/metrics.csv', index=False)
print('CLIP-IQA эталонных HR:', {k: round(v, 3) for k, v in hr_iqa.items()},
      '| в квадрате ROI:', {k: round(v, 3) for k, v in hr_iqa_in.items()})
```

## Таблицы
`bicubic` — бикубический апскейл LR; `ones` — SUPIR только по LR (M≡1); `combined` — SUPIR внутри маски с HR вне её.

```python
METRIC_COLS = ['psnr_in', 'lpips_in', 'psnr_edge', 'lpips_edge', 'clipiqa_in', 'psnr_full', 'lpips_full', 'clipiqa']
summary = (df.groupby(['degradation', 'mask'])[METRIC_COLS].mean()
             .reindex(pd.MultiIndex.from_product([DEGRADATIONS, TABLE_KINDS], names=['degradation', 'mask'])))
summary_all = df.groupby('mask')[METRIC_COLS].mean().reindex(TABLE_KINDS)
summary.to_csv(f'{OUT_DIR}/summary.csv')
summary_all.to_csv(f'{OUT_DIR}/summary_all.csv')
display(summary.round(3))
display(summary_all.round(3))
```

## Визуальное сравнение

```python
def overlay(img_u8, mask_lr):
    m = cv2.resize(mask_lr, img_u8.shape[1::-1], interpolation=cv2.INTER_NEAREST).astype(bool)
    cnt, _ = cv2.findContours(m.astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    out = img_u8.copy()
    cv2.drawContours(out, cnt, -1, (255, 64, 0), 2)
    return out

for s in samples:
    key = f"{s['img_id']}_{s['deg']}"
    imgs = {k: np.array(Image.open(f'{OUT_DIR}/outputs/{key}_{k}.png')) for k in TABLE_KINDS}
    panels = [('LR (nearest x4)', np.kron(s['lr'], np.ones((SCALE, SCALE, 1), np.uint8))),
              ('бикубика', imgs['bicubic']),
              ('M≡1: только LR', overlay(imgs['ones'], ROI)),
              ('combined: HR вне маски', overlay(imgs['combined'], MASKS['combined'])),
              ('HR', s['hr'])]
    fig, axes = plt.subplots(1, len(panels), figsize=(3.2 * len(panels), 3.4))
    for a, (t, im) in zip(axes, panels):
        a.imshow(im); a.set_title(t); a.axis('off')
    fig.suptitle(key)
    plt.tight_layout(); plt.savefig(f'{OUT_DIR}/grids/{key}.jpg', dpi=100, pil_kwargs={'quality': 90}); plt.show()

# Крупный план области маски: только LR vs LR + HR-контекст (все деградации одного изображения)
y0, y1, x0, x1 = ROI_BOX
for img_id in ids:
    row = [s for s in samples if s['img_id'] == img_id]
    fig, axes = plt.subplots(len(row), 4, figsize=(12, 3 * len(row)), squeeze=False)
    for r, s in enumerate(row):
        key = f"{s['img_id']}_{s['deg']}"
        crops = [('бикубика', f'{OUT_DIR}/outputs/{key}_bicubic.png'), ('M≡1: только LR', f'{OUT_DIR}/outputs/{key}_ones.png'),
                 ('combined: LR + HR вне маски', f'{OUT_DIR}/outputs/{key}_combined.png')]
        for c, (t, p) in enumerate(crops):
            axes[r, c].imshow(np.array(Image.open(p))[y0:y1, x0:x1]); axes[r, c].set_title(t if r == 0 else '')
        axes[r, 3].imshow(s['hr'][y0:y1, x0:x1]); axes[r, 3].set_title('HR' if r == 0 else '')
        axes[r, 0].set_ylabel(s['deg'])
        for a in axes[r]:
            a.set_xticks([]); a.set_yticks([])
    plt.tight_layout(); plt.savefig(f'{OUT_DIR}/zoom/{img_id}.jpg', dpi=100, pil_kwargs={'quality': 90}); plt.show()
```

## Абляция: на скольких шагах опираться на HR
Для маски `combined` меняется доля шагов `blend_frac`, на которых латент вне маски заменяется зашумлённым HR:
* `1.0` — основной эксперимент (берётся из него);
* `0.7`, `0.4` — смешивание только на первых 70 % / 40 % шагов;
* `0.0` — «вставка»: выход SUPIR при `M ≡ 1` (только LR), вклеенный в HR по размытой маске — HR-контекст
  во время диффузии не используется.

Метрики внутри ROI показывают качество восстановления области, метрики в полосе ±`EDGE_PX` px — качество шва.

```python
ABL_COLS = ['psnr_in', 'lpips_in', 'psnr_edge', 'lpips_edge', 'clipiqa_in']
frac_name = lambda f: 'вставка' if f == 0 else f'{f:.1f}'
abl_path = lambda key, f: (f'{OUT_DIR}/outputs/{key}_combined.png' if f == 1.0
                           else f'{OUT_DIR}/ablation/{key}_f{round(f * 100):03d}.png')
abl_rows = []
for si, s in enumerate(samples):
    key = f"{s['img_id']}_{s['deg']}"
    hr_t = to_tensor(s['hr'])
    roi_pix = mask_to_pixel(ROI, *hr_t.shape[-2:])
    for frac in ABLATION_FRACS:
        t0 = time.time()
        if frac == 1.0:
            y = main_out[(key, 'combined')].to(AE_DEVICE)
        elif frac == 0.0:
            m_soft = feather(mask_to_pixel(MASKS['combined'], *hr_t.shape[-2:]))
            y = (m_soft * main_out[(key, 'ones')].to(AE_DEVICE) + (1 - m_soft) * hr_t).clamp(-1, 1)
        else:
            y = sr_inpaint(s['lr'], MASKS['combined'], s['hr'], blend_frac=frac, **SAMPLER_KW)
        dt = time.time() - t0
        if frac != 1.0:   # 1.0 — это outputs/{key}_combined.png
            Image.fromarray(to_uint8(y)).save(abl_path(key, frac))
        abl_rows.append(dict(img_id=s['img_id'], degradation=s['deg'], blend_frac=frac, time_s=dt,
                             **compute_metrics(y, hr_t, roi_pix)))
        r = abl_rows[-1]
        print(f'[{si + 1}/{len(samples)}] {key:<16} frac={frac_name(frac):<8} {dt:5.1f}s  '
              f'LPIPS in {r["lpips_in"]:.3f}  PSNR in {r["psnr_in"]:.2f}  PSNR edge {r["psnr_edge"]:.2f}')
    torch.cuda.empty_cache()

abl = pd.DataFrame(abl_rows)
abl.to_csv(f'{OUT_DIR}/ablation/metrics.csv', index=False)
abl_sum = abl.groupby(['blend_frac', 'degradation'])[ABL_COLS].mean()
abl_all = abl.groupby('blend_frac')[ABL_COLS].mean().sort_index(ascending=False)
abl_all.to_csv(f'{OUT_DIR}/ablation/summary.csv')
display(abl_all.round(3))

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, (col, title) in zip(axes, [('lpips_in', 'LPIPS внутри маски (↓)'), ('psnr_in', 'PSNR внутри маски, дБ (↑)'),
                                   ('psnr_edge', f'PSNR у границы ±{EDGE_PX}px, дБ (↑)')]):
    for d in DEGRADATIONS:
        ax.plot(ABLATION_FRACS, [abl_sum.loc[(f, d), col] for f in ABLATION_FRACS], 'o-', label=d, alpha=0.8)
    ax.plot(ABLATION_FRACS, [abl_all.loc[f, col] for f in ABLATION_FRACS], 'ks-', lw=2, label='среднее')
    if col in ('lpips_in', 'psnr_in'):
        ax.axhline(summary_all.loc['ones', col], color=KIND_COLORS['ones'], ls='--', lw=1, label='M≡1 (среднее)')
        ax.axhline(summary_all.loc['bicubic', col], color=KIND_COLORS['bicubic'], ls='--', lw=1, label='бикубика (среднее)')
    ax.set_xticks(ABLATION_FRACS); ax.set_xticklabels([frac_name(f) for f in ABLATION_FRACS])
    ax.set_xlabel('доля шагов с латентным смешиванием'); ax.set_title(title); ax.invert_xaxis()
fig.legend(*axes[0].get_legend_handles_labels(), loc='lower center', ncol=7, frameon=False, fontsize=9)
plt.tight_layout(rect=(0, 0.07, 1, 1)); plt.savefig(f'{OUT_DIR}/ablation/chart.png', dpi=120); plt.show()

abl_examples = [s for s in samples if s['deg'] == DEGRADATIONS[-1]]
fig, axes = plt.subplots(len(abl_examples), len(ABLATION_FRACS) + 2,
                         figsize=(2.8 * (len(ABLATION_FRACS) + 2), 2.9 * len(abl_examples)), squeeze=False)
for r, s in enumerate(abl_examples):
    key = f"{s['img_id']}_{s['deg']}"
    panels = [('бикубика', np.array(Image.open(f'{OUT_DIR}/outputs/{key}_bicubic.png')))]
    panels += [(f'blend_frac={frac_name(f)}',
                overlay(np.array(Image.open(abl_path(key, f))), MASKS['combined']))
               for f in ABLATION_FRACS]
    panels += [('HR', s['hr'])]
    for c, (t, im) in enumerate(panels):
        axes[r, c].imshow(im); axes[r, c].set_title(t if r == 0 else ''); axes[r, c].axis('off')
    axes[r, 0].set_title(f'{key}\nбикубика', fontsize=9)
plt.tight_layout(); plt.savefig(f'{OUT_DIR}/ablation/examples.jpg', dpi=100, pil_kwargs={'quality': 90}); plt.show()
```
